# 2026 期末试卷

## 1

(1) 拓扑空间的定义

(2) 开映射

(3) 商拓扑下集合 $[0,1]\setminus\{0,1\}$ 的拓扑形态说明

(4) 乌雷松引理

## 2

(1) 设 $X=\mathbb R^2\setminus\{(0,0)\}$，判断 $X$ 是否连通、是否道路连通，并说明理由。

(2) 判断映射 $f:\mathbb R^2\to\mathbb R$ 是否可能为同胚，说明原因。

## 3

设 $\overline{A}=(0,1]\cup\{2\}\cup(3,4)\subset\mathbb R$，$\mathbb R$ 赋予通常欧氏拓扑，求集合 $\overline{A}$ 的内部 $\mathring{\overline{A}}$ 与闭包 $\overline{\overline{A}}$。

## 4

设拓扑空间 $X$ 紧致，$Y$ 为豪斯多夫空间，映射 $f:X\to Y$ 连续且为双射。

(1) 任取 $X$ 的闭子集 $\overline{A}$，求证 $f(\overline{A})$ 是 $Y$ 中的闭集；

(2) 求证 $f$ 是同胚映射；

(3) 举出例子说明：仅连续双射无法推出映射是同胚。

## 5

(1) 写出积拓扑的一组拓扑基；

(2) 投影映射 $\pi_X:X\times Y\to X,\ \pi_Y:X\times Y\to Y$，证明二者既是连续映射也是开映射；

(3) 若拓扑空间 $X,Y$ 均为正则空间，证明乘积空间 $X\times Y$ 也是正则空间。

## 6

设 $f:X\to Y$ 为连续满射，$X$ 紧致，证明 $Y$ 紧致。

## 7

集合 $X=\{a,b,c\}$，写出其上全部9个拓扑结构；并针对每个拓扑，分别判定：是否豪斯多夫、是否紧致、是否连通。

## 8

(1) 用 $\varepsilon$-$\delta$ 语言写出度量空间之间映射连续的定义；

(2) 简述度量空间中柯西列与完备空间的定义以及二者的关系；

(3) 证明：紧致度量空间中的任意无穷序列必存在收敛子列。
